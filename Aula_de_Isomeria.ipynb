```python
import py3Dmol
from rdkit import Chem
from rdkit.Chem import AllChem

def mol_to_3D(mol, addHs=True):
    """Gera coordenadas 3D para a molécula"""
    mol = Chem.AddHs(mol) if addHs else mol
    AllChem.EmbedMolecule(mol, AllChem.ETKDG())
    AllChem.UFFOptimizeMolecule(mol)  # Otimiza a estrutura
    return mol

def translate_molecule(mol, x_offset=0, y_offset=0, z_offset=0):
    """Desloca a molécula no espaço"""
    conf = mol.GetConformer()
    for i in range(mol.GetNumAtoms()):
        pos = list(conf.GetAtomPosition(i))
        conf.SetAtomPosition(i, (pos[0] + x_offset, pos[1] + y_offset, pos[2] + z_offset))

def draw_ball_and_stick(mol1, mol2):
    """Plota duas moléculas em 3D no modelo bolas e tacos"""
    viewer = py3Dmol.view(width=1000, height=1000)

    # Desloca a segunda molécula no eixo X para evitar sobreposição
    translate_molecule(mol2, x_offset=15.0)

    for i, mol in enumerate([mol1, mol2]):
        mol_block = Chem.MolToMolBlock(mol)
        viewer.addModel(mol_block, "mol")
        viewer.setStyle({"model": i}, {
            "stick": {"radius": 0.2},  # Ligações mais espessas
            "sphere": {"scale": 0.3}  # Átomos visíveis como esferas pequenas
        })

    viewer.zoomTo()
    return viewer.show()

butane = Chem.MolFromSmiles("CC")
isobutane = Chem.MolFromSmiles("CC")

butane_3D = mol_to_3D(butane)
isobutane_3D = mol_to_3D(isobutane)

draw_ball_and_stick(butane_3D, isobutane_3D)
```
